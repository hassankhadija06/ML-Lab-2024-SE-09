##  Task 01: 7 to the power of 4

**Problem:** What is 7 raised to the power of 4?

**Approach:** Python's exponent operator is `**`, so `7 ** 4` means 7 × 7 × 7 × 7.

```python
# Task 01: exponentiation with the ** operator
base = 7
exponent = 4

result = base ** exponent

print(f"{base} to the power of {exponent} = {result}")
```

##  Task 04: Indexing a nested list

**Problem:** Use indexing to grab `'hello'` from the nested list `lst`.

**Approach:** Go one level deeper at a time. Each index peels off one layer of brackets.

| Step | Expression | Result |
|:--:|:--|:--|
| 1 | `lst[3]` | `[5, [100, 200, ['hello']], 23, 11]` |
| 2 | `lst[3][1]` | `[100, 200, ['hello']]` |
| 3 | `lst[3][1][2]` | `['hello']` |
| 4 | `lst[3][1][2][0]` | `'hello'`  |

```python
# Task 04: nested list indexing
lst = [1, 2, [3, 4], [5, [100, 200, ['hello']], 23, 11], 1, 7]

target = lst[3][1][2][0]   # 4th item -> 2nd item -> 3rd item -> 1st item

print(target)
```

##  Task 05: Indexing a nested dictionary

**Problem:** Use indexing to grab `'hello'` from the nested dictionary `d`.

**Approach:** Use **keys** for dictionaries and **positions** for lists, one layer at a time.

| Step | Expression | Result |
|:--:|:--|:--|
| 1 | `d['k1']` | `[1, 2, 3, {'tricky': [...]}]` |
| 2 | `d['k1'][3]` | `{'tricky': ['oh', 'man', 'inception', {...}]}` |
| 3 | `d['k1'][3]['tricky']` | `['oh', 'man', 'inception', {'target': [...]}]` |
| 4 | `d['k1'][3]['tricky'][3]` | `{'target': [1, 2, 3, 'hello']}` |
| 5 | `d['k1'][3]['tricky'][3]['target']` | `[1, 2, 3, 'hello']` |
| 6 | `d['k1'][3]['tricky'][3]['target'][3]` | `'hello'`  |

```python
# Task 05: nested dictionary indexing
d = {'k1': [1, 2, 3, {'tricky': ['oh', 'man', 'inception', {'target': [1, 2, 3, 'hello']}]}]}

target = d['k1'][3]['tricky'][3]['target'][3]   # key -> index -> key -> index -> key -> index

print(target)
```

##  Task 06: Tuple vs. list

**Problem:** What is the main difference between a tuple and a list?

### Answer
> **The main difference is mutability: a list is *mutable* (can be changed after creation), while a tuple is *immutable* (cannot be changed).**

| Feature | List | Tuple |
|:--|:--|:--|
| **Mutability** | Mutable  | Immutable  |
| **Syntax** | Square brackets `[ ]` | Round brackets `( )` |
| **Item assignment** | `my_list[0] = x` works | `my_tuple[0] = x` raises `TypeError` |
| **Typical use** | Collections that change | Fixed data (e.g. coordinates) |
| **Common points** | Ordered, indexable, can hold mixed data types | Ordered, indexable, can hold mixed data types |

The demonstration below proves the difference in code.

```python
# Task 06: demonstrating mutability
my_list = [1, 2, 3, 5]
my_tuple = (1, 2, 3, 5)

# Lists can be modified in place
my_list[3] = 4
print("Modified list :", my_list)

# Tuples cannot be modified, so this raises a TypeError
try:
    my_tuple[3] = 4
except TypeError as error:
    print("Tuple error   :", error)
```

##  Task 07: Extract the domain from an email

**Problem:** Write a function that grabs the domain from `"user@domain.com"`.

**Approach:** Split the email on `'@'`. The part **after** the `@` is at index `1`, so `[1]` selects it.

```python
# Task 07: function to extract the domain of an email address
def get_domain(email):
    """Return the domain part of an email address (the text after '@')."""
    return email.split('@')[1]


print(get_domain("user@domain.com"))
```

##  Task 08: Membership test (case-insensitive)

**Problem:** Write a function that returns `True` if the word `"dog"` is in a string, ignoring case.

**Approach:** Convert the text to lowercase with `.lower()`, then use the `in` operator. The expression already evaluates to a `bool`, so it can be returned directly.

```python
# Task 08: case-insensitive membership test
def find_dog(text):
    """Return True if the word 'dog' appears in text (case-insensitive), else False."""
    return "dog" in text.lower()


print(find_dog("A quick brown fox jumps over the lazy dog"))     # True
print(find_dog("Dog is my favourite pet"))  # True  (capital D is handled)
print(find_dog("I own a cat"))              # False
```

##  Task 09: Count occurrences (case-insensitive)

**Problem:** Write a function that counts how many times `"dog"` occurs in a string, ignoring case.

**Approach:** Lowercase the text, then use the built-in `str.count()`.

```python
# Task 09: case-insensitive count
def count_dog(text):
    """Return how many times 'dog' occurs in text (case-insensitive)."""
    return text.lower().count("dog")


print(count_dog("This dog runs faster than the other dog"))                       # 2
print(count_dog("Dogs are loyal. My DOG loves other dogs."))                      # 3
print(count_dog("I only like cats"))                                              # 0
```

##  Task 10: `caught_speeding(speed, is_birthday)`

**Problem:** Return a ticket level based on speed. On your birthday, **5 mph extra** is allowed.

| Speed | Result |
|:--|:--|
| ≤ 60 | `"No Ticket"` |
| 61 – 80 | `"Small Ticket"` |
| ≥ 81 | `"Big Ticket"` |

**Approach:**
1. If it is the driver's birthday, subtract 5 from the speed (equivalent to allowing 5 mph more).
2. Compare the adjusted speed against the limits using `if / elif / else`.
3. **Return** the result (so it can be reused and tested) and print it when calling.

>  **Edge case:** the birthday allowance must shift *every* boundary, e.g. speed 65 on a birthday → 60 → `"No Ticket"`, and speed 85 on a birthday → 80 → `"Small Ticket"`.

```python
# Task 10: speeding ticket logic
def caught_speeding(speed, is_birthday):
    """Return 'No Ticket', 'Small Ticket' or 'Big Ticket' for the given speed.

    On a birthday the driver is allowed 5 mph more, so the speed is reduced by 5.
    """
    if is_birthday:
        speed -= 5            # 5 mph extra allowed on birthday

    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"


# Test cases: (speed, is_birthday, expected)
test_cases = [
    (60, False, "No Ticket"),      # exactly on the limit
    (65, True,  "No Ticket"),      # birthday edge case: 65 -> 60
    (61, False, "Small Ticket"),
    (85, True,  "Small Ticket"),   # birthday edge case: 85 -> 80
    (81, False, "Big Ticket"),
    (86, True,  "Big Ticket"),     # birthday edge case: 86 -> 81
]

for speed, is_birthday, expected in test_cases:
    actual = caught_speeding(speed, is_birthday)
    status = "PASS" if actual == expected else "FAIL"
    print(f"{status} | speed={speed:>3}, birthday={str(is_birthday):<5} -> {actual}")
```

### Bonus 1: Count "dog" even with punctuation

**Problem:** Make `count_dog` also count `"dog,"` and `"dog!"` (punctuation attached to the word).

**Approach:**
1. Convert the text to lowercase.
2. Split it into words.
3. Strip punctuation from both ends of each word with `str.strip(string.punctuation)`.
4. Count the words that are exactly `"dog"`.

> **Note:** this version counts the *whole word* only, so `"dogs"` and `"hotdog"` are not counted. The simpler `count_dog` from Task 09 counts any occurrence of the letters "dog".

```python
# Bonus 1: count "dog" even when it has punctuation attached
import string


def count_dog_punctuation(text):
    """Count the word 'dog' (case-insensitive), ignoring punctuation around it."""
    count = 0
    for word in text.lower().split():
        if word.strip(string.punctuation) == "dog":
            count += 1
    return count


sample = "Dog, my dog! A dog. Dogs, hotdog, DOG?"

print(count_dog_punctuation("I saw a dog, a dog! and a dog."))   # 3
print(count_dog_punctuation(sample))                             # 4
print(count_dog_punctuation("No pets here"))                     # 0
```

### Bonus 2: Palindrome check using slicing

**Problem:** Write `is_palindrome(word)` using slicing `word[::-1]`.

**Approach:** The slice `[::-1]` reads a string backwards. A palindrome reads the same forwards and backwards, so the word is compared with its reverse. The word is lowercased first so that `"Level"` also works.

```python
# Bonus 2: palindrome check with slicing
def is_palindrome(word):
    """Return True if word reads the same forwards and backwards (case-insensitive)."""
    word = word.lower()
    return word == word[::-1]


print(is_palindrome("madam"))     # True
print(is_palindrome("Level"))     # True
print(is_palindrome("python"))    # False
```

### Bonus 3: Students as a list of dictionaries

**Problem:** Store 5 students as a list of dictionaries and print the topper's name.

**Approach:** Each student is a dictionary with a `"name"` and `"marks"` key. The built-in `max()` with `key=` picks the dictionary with the highest marks, and then `["name"]` gives the topper's name.

```python
# Bonus 3: list of dictionaries and finding the topper
students = [
    {"name": "Ayesha", "marks": 82},
    {"name": "Bilal",  "marks": 91},
    {"name": "Hira",   "marks": 76},
    {"name": "Usman",  "marks": 88},
    {"name": "Zainab", "marks": 69},
]

for student in students:
    print(f"{student['name']:<8} {student['marks']}")

topper = max(students, key=lambda student: student["marks"])

print("\nTopper:", topper["name"])
```
